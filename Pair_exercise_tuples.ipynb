```python
# we can combine tuples, since we can put anything we want in a typle

b=a,3,4,5

b
```

```python
len(b)
```

```python
#tuple unpacking

w,x,y,z=b

print(w)
print(x)
print(y)
print(z)
```

# Tuples and comparisons

We can use tuples in comparisons (>,<, ==, >=, etc)

If we compare 2 tuples each with 3 items, the comparison is done using the first item,  if the first items are equal, the comparison is
applied to the second and so on

```python
a= "beauty",2,"afghan"
b="beauty",2,"throw-pillow"

```

```python
a>b
```

```python
b>a
```

```python
c=1980,10,1
d=1980,12,2

c<d
```

```python
a==b
```
